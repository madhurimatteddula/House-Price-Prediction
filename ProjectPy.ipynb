```python
df.drop(columns=['w','zipcode'],inplace=True)
```

```python
df['date']=df['date'].str.slice(0,8)
```

```python
df['date']=pd.to_datetime(df['date'])
```

```python
df['Year']=df['date'].dt.year
```

```python
df.head(2)
```

```python
df['House_age']=df['Year']-df['yr_built']
```

```python
df.head(2)
```

```python
df['Renovation']=df['yr_renovated'].apply(lambda x: 1 if x>0 else 0)
```

```python
df.head(8)
```

```python
col_drop=['date','yr_built','yr_renovated','Year']
df.drop(columns=col_drop,inplace=True)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(12,12))
sns.heatmap(df.corr(numeric_only=True),cmap='RdBu',annot=True,fmt='.2f')
```

```python
df.drop('sqft_above',axis=1,inplace=True)
```

```python
df.columns
```

```python
x=df.drop('price',axis=1)
y=df['price']
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.fit_transform(x_test)
x_train_scaled
```

```python
from sklearn.linear_model import LinearRegression
reg=LinearRegression()
reg.fit(x_train_scaled,y_train)
pred=reg.predict(x_test_scaled)
pred
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,mean_absolute_percentage_error,r2_score
mae=mean_absolute_error(y_test,pred)
mse=mean_squared_error(y_test,pred)
mapr=mean_absolute_percentage_error(y_test,pred)
r2=r2_score(y_test,pred)
print("mean_absolute_error",mae)
print("mean_squared_error",mse)
print("mean_absolute_percentage_error",mapr)
print("r2_score",r2)
```

```python
reg_train_score=reg.score(x_train_scaled,y_train)
reg_test_score=reg.score(x_test_scaled,y_test)
gap = reg_train_score-reg_test_score
print("train score :",reg_train_score)
print("test_score :",reg_test_score)
print("gap",gap)
```

```python
from sklearn .tree import DecisionTreeRegressor
tree=DecisionTreeRegressor(max_depth=3)
tree.fit(x_train_scaled,y_train)
y_pred=tree.predict(x_test_scaled)
y_pred
```

```python
from sklearn.metrics import mean_absolute_percentage_error,r2_score
mapr=mean_absolute_percentage_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)
print("mean_absolute_percentage_error",mapr)
print("r2_score",r2)
```

```python
from sklearn.ensemble import RandomForestRegressor
Model=RandomForestRegressor(n_estimators=100,random_state=42)
Model.fit(x_train_scaled,y_train)
rf=Model.predict(x_test_scaled)
rf_train=Model.score(x_train_scaled,y_train)
rf_test=Model.score(x_test_scaled,y_test)
gap=rf_train-rf_test
print("train score :",rf_train)
print("test_score :",rf_test)
print("gap",gap)
```

```python
from sklearn.metrics import r2_score,mean_absolute_percentage_error
mape=mean_absolute_percentage_error(y_test,rf)
r2=r2_score(y_test,rf)
print("mean_absolute_percentage_error",mape)
print("r2_score",r2)
```

```python
rf=RandomForestRegressor(
    n_estimators=300,
    max_depth=15,
    min_samples_leaf=5,
    min_samples_split=10,
    max_features='sqrt',
    random_state=42
)
rf.fit(x_train_scaled,y_train)
random_forest=rf.predict(x_test_scaled)
rf1_train=rf.score(x_train_scaled,y_train)
rf2_test=rf.score(x_test_scaled,y_test)
gap=rf1_train-rf2_test

print("train score :",rf1_train)
print("test_score :",rf2_test)
print("gap",gap)
```

```python
from sklearn.metrics import mean_absolute_percentage_error,r2_score
import numpy as np
per_mae=mean_absolute_percentage_error(y_test,random_forest)
r2_score= r2_score(y_test,random_forest)
print("r2_score",r2_score)
print("mean_absolute_percentage_error",per_mae)
```

```python
from xgboost import XGBRegressor
xgb=XGBRegressor(
    n_estimators=1000,
    learning_rate=0.05,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    n_jobs=-1
)
xgb.fit(x_train,y_train)
xgb_pred=xgb.predict(x_test)
xgb_train=xgb.score(x_train,y_train)
xgb_test=xgb.score(x_test,y_test)
gap=xgb_train-xgb_test

print("train score :",xgb_train)
print("test_score :",xgb_test)
print("gap",gap)
```

```python
from sklearn.metrics import mean_absolute_percentage_error,r2_score
per_mae=mean_absolute_percentage_error(y_test,xgb_pred)
r2_score= r2_score(y_test,xgb_pred)
print("r2_score",r2_score)
print("mean_absolute_percentage_error",per_mae)
```
